```python
import pandas as pd
import numpy as np
import random
random.seed(42)
```

```python
final_list = []
for x in range(1, 30*24 + 1,1):
    hour = x%24
    if hour <= 18 and hour >= 11:
        if final_list[x-2]==5000:
            value = random.choice([5000, random.uniform(750, 1750)])
        else:
            value = 5000
    else:
        value = 5000
    final_list.append(value)
```

```python
pd.DataFrame(final_list,columns = ['power_limit']).to_csv('data/TOU_IML.csv')
```
