```python
from sklearn.ensemble import RandomForestRegressor
import pandas as pd
import pickle
import numpy as np
```

```python
X_1 = pd.DataFrame([30,29,28,27,26,25,24,23])

y_1 = pd.DataFrame([200, 190,180,170,160,150,140,130]).values.ravel()

regr_1 = RandomForestRegressor(max_depth=2, random_state=0)
regr_1.fit(X_1, y_1)

filename_1 = 'model/data-driven_building_energy_model_1.sav'
pickle.dump(regr_1, open(filename_1, 'wb'))
```

```python
X_2 = pd.DataFrame([30,29,28,27,26,25,24,23])

y_2 = pd.DataFrame([209, 199,189,179,169,159,149,139]).values.ravel()

regr_2 = RandomForestRegressor(max_depth=2, random_state=0)
regr_2.fit(X_2, y_2)

filename_2 = 'model/data-driven_building_energy_model_2.sav'
pickle.dump(regr_2, open(filename_2, 'wb'))
```
